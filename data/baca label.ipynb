```python
import os
import glob
from collections import defaultdict

DATASET_PATH = "./data"

CATEGORIES = ["Beauty", "F&B", "Gas", "Groceries",
              "Health", "HouseHold", "Lifestyle", "Listrik", "Transport"]

SPLITS = ["train", "valid", "test"]
image_exts = ["*.jpg", "*.jpeg", "*.png", "*.webp"]

print("=" * 75)
print("  ANALISIS DATASET YOLO per KATEGORI + SPLIT")
print("=" * 75)

grand_total_img = 0
grand_total_lbl = 0
summary = []

for cat in CATEGORIES:
    cat_path = os.path.join(DATASET_PATH, cat)

    if not os.path.exists(cat_path):
        print(f"\n❌ [{cat}] folder tidak ditemukan!")
        continue

    print(f"\n📂 {cat}")
    cat_total_img = 0
    cat_total_lbl = 0

    for split in SPLITS:
        img_dir = os.path.join(cat_path, split, "images")
        lbl_dir = os.path.join(cat_path, split, "labels")

        # Kalau tidak ada subfolder split, cek langsung di cat_path
        if not os.path.exists(img_dir):
            img_dir = os.path.join(cat_path, "images")
            lbl_dir = os.path.join(cat_path, "labels")
            split_label = "flat (no split)"
        else:
            split_label = split

        images = []
        for ext in image_exts:
            images += glob.glob(os.path.join(img_dir, ext))

        labels = glob.glob(os.path.join(lbl_dir, "*.txt")) if os.path.exists(lbl_dir) else []

        if images or labels:
            match = "✅" if len(images) == len(labels) else "⚠️ MISMATCH"
            print(f"   [{split_label:6s}]  gambar: {len(images):4d}  |  label: {len(labels):4d}  {match}")
            cat_total_img += len(images)
            cat_total_lbl += len(labels)

        # Hindari duplikasi kalau flat
        if split_label == "flat (no split)":
            break

    print(f"   {'SUBTOTAL':8s}  gambar: {cat_total_img:4d}  |  label: {cat_total_lbl:4d}")
    summary.append((cat, cat_total_img, cat_total_lbl))
    grand_total_img += cat_total_img
    grand_total_lbl += cat_total_lbl

# ── Sample isi label YOLO ──────────────────────────────────────
print("\n" + "=" * 75)
print("  SAMPLE ISI FILE LABEL (YOLO format check)")
print("=" * 75)
sample_txts = glob.glob(os.path.join(DATASET_PATH, "**", "labels", "*.txt"), recursive=True)
if sample_txts:
    with open(sample_txts[0]) as f:
        lines = f.readlines()
    print(f"  File : {sample_txts[0]}")
    print(f"  Isi  :")
    for line in lines[:5]:
        parts = line.strip().split()
        print(f"    class_id={parts[0]}  cx={parts[1]}  cy={parts[2]}  w={parts[3]}  h={parts[4]}")
else:
    print("  Tidak ada file .txt ditemukan di subfolder labels/")

# ── Cek classes.txt / data.yaml ───────────────────────────────
print("\n" + "=" * 75)
print("  CEK NAMA KELAS (classes.txt / data.yaml)")
print("=" * 75)
found_classes = False
for fname in ["classes.txt", "obj.names", "data.yaml"]:
    fpath = os.path.join(DATASET_PATH, fname)
    if os.path.exists(fpath):
        with open(fpath) as f:
            print(f"  File: {fname}\n  Isi :\n{f.read()}")
        found_classes = True
        break

# Cari data.yaml di dalam kategori
if not found_classes:
    yamls = glob.glob(os.path.join(DATASET_PATH, "**/*.yaml"), recursive=True)
    if yamls:
        with open(yamls[0]) as f:
            print(f"  File: {yamls[0]}\n{f.read()}")
    else:
        print("  ⚠️  Tidak ada file nama kelas — class_id di label mungkin perlu dikonfirmasi manual")

# ── Grand Total ───────────────────────────────────────────────
print("\n" + "=" * 75)
print(f"  {'KATEGORI':<15} {'GAMBAR':>8} {'LABEL':>8} {'STATUS':>12}")
print("  " + "-" * 45)
for cat, img, lbl in summary:
    status = "✅ OK" if img == lbl else "⚠️ MISMATCH"
    print(f"  {cat:<15} {img:>8} {lbl:>8} {status:>12}")
print("  " + "-" * 45)
print(f"  {'TOTAL':<15} {grand_total_img:>8} {grand_total_lbl:>8}")
print("=" * 75)
```
